```python
# 1) 프로젝트 루트 계산 & src 임포트 경로 추가
import sys
from pathlib import Path

# 이 노트북이 notebooks/ 디렉터리에 있으므로,
# 부모 디렉터리(=deepfake-adapter) 를 프로젝트 루트로 지정
project_root = Path().resolve().parents[0]
sys.path.insert(0, str(project_root))

# 확인
print("Project root:", project_root)
```

## 1. inference.py 기능 검증

이 셀은 `scripts/inference.py` 를 호출해 정상적으로 프레임 추출 → 예측 → JSON 저장이 되는지 확인합니다.

```python
import subprocess
from pathlib import Path

# deepfake-adapter 프로젝트 루트
proj = Path().resolve().parents[1] / "deepfake-adapter"

# 1) inference.py 실행
cmd = [
    "python", str(proj/"scripts"/"inference.py")
]
res = subprocess.run(
    cmd,
    cwd=str(proj),            # 여기서도 deepfake-adapter 폴더를 cwd 로!
    capture_output=True,
    text=True
)
print(res.stdout)
print(res.stderr)

# 2) 결과 JSON 파일 확인
out_json = proj/"outputs"/"predictions"/"sample_video.json"
print("Exists?", out_json.exists())
```

## 2️⃣ xai_inference.py 기능 검증

`scripts/xai_inference.py` 를 호출해 saliency 맵을 생성하고, `scripts/xai_saliency/` 에 이미지가 잘 저장되는지 확인합니다.


```python
import subprocess, os
from pathlib import Path

proj = Path().resolve().parents[1] / "deepfake-adapter"
script = proj/"scripts"/"xai_inference.py"

cmd = [
    "python", str(script),
    "--frames-dir", str(proj/"scripts"/"temp_frames_infer"),
    "--output-dir", str(proj/"scripts"/"xai_saliency"),
    "--target-class", "1",
    "--batch-size", "32",
    "--num-workers", "4"
]
res = subprocess.run(cmd, cwd=str(proj), capture_output=True, text=True)
print(res.stdout)
print(res.stderr)

sal_dir = proj/"scripts"/"xai_saliency"
files = sorted(os.listdir(sal_dir)) if sal_dir.exists() else []
print(f"{len(files)} saliency files:", files[:5])
```
